Atividade 1
Faça na linguagem Python uma função que verifica se um número é perfeito.

Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6 = 1 + 2 + 3).

````python
def is_perfect(n: int) -> bool:
    if n < 1:
        return False

    sum_divisors = 0
    for i in range(1, n):
        if n % i == 0:
            sum_divisors += i

    return sum_divisors == n


def main():
    assert is_perfect(6) == True
    assert is_perfect(7) == False
    assert is_perfect(-1) == False


if __name__ == "__main__":
    main()
    
````

## Atividade 2
Faça na linguagem Python que calcula o fatorial de um número;
O fatorial de um número é o produto do número por todos os antecessores positivos.


````python
def fatorial(n):
    """Retorna o fatorial de n."""
    if n < 0:
        return "Número inválido"
    if n == 0:
        return 1

    resultado = 1
    for i in range(1, n + 1):
        resultado *= i

    return resultado


# Exemplos
for numero in [0, 1, 5, 7]:
    print(f"{numero}! = {fatorial(numero)}")
````
